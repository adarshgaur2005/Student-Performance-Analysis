```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#load csv file
df = pd.read_csv("StudentsPerformance.csv")
df.head()
```

```python
#to understand dataset
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns)

print("\nMissing Values:")
print(df.isnull().sum())

df.describe()
```

```python
# calculate average marks
subjects = ["math score", "reading score", "writing score"]

print("Average Marks:")
print(df[subjects].mean().round(2))
```

```python
# plot subject-wise graph
df[subjects].mean().plot(
    kind="bar",
    title="Average Student Performance",
    ylabel="Average Marks",
    xlabel="Subjects"
)

plt.ylim(0, 100)
plt.show()
```

```python
# boys vs girls performance
df.groupby("gender")[subjects].mean().plot(
    kind="bar",
    title="Performance by Gender"
)

plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.show()
```

```python
# effect of test preparation
df.groupby("test preparation course")[subjects].mean().plot(
    kind="bar",
    title="Test Preparation vs Performance"
)

plt.ylabel("Average Marks")
plt.ylim(0, 100)
plt.xticks(rotation=0)
plt.show()
```

```python
KEY FINDINGS:
1.The subject with the highest average marks is Rading.

2.The subject with the lowest average marks is Maths.

3.Students who completed test preparation had acheived Higher average scores compared with students who did not.

4.Female students scored higher on average in Reading and Writing, while male students scored higher on average in Mathematics.

5.The correlation analysis shows a positive relationship between Mathematics, Reading, and Writing scores.
```
